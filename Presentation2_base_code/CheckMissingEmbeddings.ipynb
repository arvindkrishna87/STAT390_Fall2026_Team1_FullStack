```python
import os
from pathlib import Path
```

```python
png_dir = Path("/projects/e32998/patches_benign_split")
pt_dir = Path("/projects/e32998/patches_varsize_pooled4096")

png_files = list(png_dir.rglob("*.png"))

missing = []
for png in png_files:
    expected_pt = pt_dir / (png.stem + ".pt")
    if not expected_pt.is_file():
        missing.append((png.name, str(expected_pt)))

print(f"Total PNGs: {len(png_files)}")
print(f"Missing embeddings: {len(missing)}")
print(f"Available embeddings: {len(png_files) - len(missing)}")
```
